# 001 — Phenotype-to-Genotype Research Framework

## Purpose

This first notebook defines the scope, standards, terminology, and long-term research direction of the Phenotype-to-Genotype project. It also goes over the background, intent, and overall sentiment of the project.

This project investigates how variation in human genotype contributes to variation in measurable phenotype, with an emphasis on:

- reproducible computational analysis
- causal biological mechanisms
- quantitative phenotype measurement
- population genetics
- environmental effects
- gene–gene interactions
- gene–environment interactions
- replication of established findings
- identification of unexplained phenotypic variation

It then investigates how to actually predict, or select for, phenotypes based on a genotype, and the methods to do so.

The initial goal is not to discover novel biology immediately. The initial goal is to build a rigorous, reproducible framework capable of reproducing known phenotype--genotype relationships, then eventually extending beyond the known ones, into the more complex unknown traits.

## Background

The author was always interested in diversity and what causes it. The author was also keenly interested in justice and equality and why certain traits seemed to be highly advantageous over others. After excelling in mendelian biology, the author developed a passion for taking control of human phenotypical outcomes by harnessing the genome. This project is a lifelong exploration of that goal, as career requirements have taken the author on a deifferent course. Despire this, the author still desires to deeply understand why and how huamn diversity occurs, and ideally acheive an understanding of how we could, in theory, alter phenotypical outcomes at the lowest genetic level in order to reduce suffering and increase equality.
